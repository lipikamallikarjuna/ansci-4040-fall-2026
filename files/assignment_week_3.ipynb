```python
# Display data with summary statistics
display(cowstats)

# Generate summary statistics
print("\n=== Dataset Summary ===")
print(f"Total rows: {len(cowstats)}")
print(f"Total columns: {len(cowstats.columns)}")
print(f"\nColumn names: {list(cowstats.columns)}")

print("\n=== Data Info ===")
print(cowstats.info())

print("\n=== Summary Statistics ===")
print(cowstats.describe())

print("\n=== Missing Values ===")
print(cowstats.isnull().sum())

print("\n=== Unique Values per Column ===")
for col in cowstats.columns:
    print(f"{col}: {cowstats[col].nunique()} unique values")
```

```python
#2.1 How many countries are in this dataset? 
# Count unique entities (countries)
num_countries = cowstats['Entity'].nunique()
print(f"Number of countries in this dataset: {num_countries}")
```

```python
#2.2 What is the time period that data was collected over?
# Find the minimum and maximum year from the Year column
min_year = cowstats['Year'].min()
max_year = cowstats['Year'].max()
print(f"Time period: {min_year} to {max_year}")
print(f"Data was collected over {max_year - min_year} years")
```

```python
#2.3 What is the mean number of cows per country?
mean_cattle_per_country = cowstats.groupby('Entity')['Cattle - Stocks (animals)'].mean()
print(f"Mean number of cows per country: {mean_cattle_per_country.mean()}")
```

```python
# 2.4 What is the median number of cows per country?
median_cattle_per_country = cowstats.groupby('Entity')['Cattle - Stocks (animals)'].median()
print(f"Median number of cows per country: {median_cattle_per_country.median()}")
```

```python
#3.1 How many more cows were there in 2024 compared to 1965?
# Filter data for 1965 and 2024
cows_1965 = cowstats[cowstats['Year'] == 1965]['Cattle - Stocks (animals)'].sum()
cows_2024 = cowstats[cowstats['Year'] == 2024]['Cattle - Stocks (animals)'].sum()

# Calculate the difference
difference = cows_2024 - cows_1965

print(f"Total cows in 1965: {cows_1965:,.0f}")
print(f"Total cows in 2024: {cows_2024:,.0f}")
print(f"Difference: {difference:,.0f} more cows in 2024 compared to 1965")
```

```python
#3.2 What is the percentage increase in cows 2024 versus 1965?
# Calculate percentage increase
percentage_increase = ((cows_2024 - cows_1965) / cows_1965) * 100

print(f"Percentage increase from 1965 to 2024: {percentage_increase:.2f}%")
```
